Question 3: An e-commerce platform has customer IDs from two campaigns:
campaign1 = {101, 102, 103, 104, 105}
campaign2 = {104, 105, 106, 107}
1. Find unique customers across both campaigns.
2. Find customers common in both.
3. Find customers exclusive to campaign1.
4. Explain how sets help in removing duplicates in real DS projects.

```python
campaign1 = {101, 102, 103, 104, 105}
print(campaign1)
```

```python
campaign2 = {104, 105, 106, 107}
print(campaign2)
```

```python
#Use the union (|) operation:
unique_customers = campaign1 | campaign2

print(unique_customers)
```

```python
#Use the intersection (&) operation:

common_customers = campaign1 & campaign2

print(common_customers)
```

```python
#Use the difference (-) operation:

exclusive_campaign1 = campaign1 - campaign2

print(exclusive_campaign1)
```

```python
#Sets automatically store only unique values, making them very useful for data cleaning and preprocessing.
customer_ids = [101, 102, 103, 101, 104, 102, 105]

unique_ids = set(customer_ids)

print(unique_ids)
```

Sets are valuable in Data Science because they:

1.Automatically remove duplicates.
2.Support efficient operations like union, intersection, and difference.
3.Help clean and analyze large datasets quickly and accurately.

Machine Learning models require data in the correct format. If values remain as strings instead of numbers, several problems can occur:

1->Mathematical operations fail
"499.99" * "5"   # Error
The model cannot perform calculations on string data.

2->Incorrect results

"5" + "10"

Output:

"510"

Instead of the expected numeric result 15.

3->Model training errors
Algorithms such as Linear Regression, Decision Trees, and Neural Networks expect numeric inputs. String values may cause training to fail.
4->Data preprocessing issues
Feature scaling, normalization, and statistical calculations require numeric data types. Incorrect types can produce errors or invalid outputs.

Example: If product prices are stored as strings, an ML model predicting sales revenue cannot calculate averages, trends, or correlations correctly, leading to inaccurate predictions or pipeline failures.

Question 5: Given:
sales = [12000, 15000, 11000, 18000, 17000]
Using a loop:
1. Calculate total revenue.
2. Count how many months had sales above 15000.
3. Find maximum sales without using built-in max().

```python
sales = [12000, 15000, 11000, 18000, 17000]
print(sales)
```

```python
total_revenue = 0

for sale in sales:
    total_revenue += sale

print("Total Revenue:", total_revenue)
```

```python
count = 0

for sale in sales:
    if sale > 15000:
        count += 1

print("Months above 15000:", count)
```

```python
maximum = sales[0]

for sale in sales:
    if sale > maximum:
        maximum = sale

print("Maximum Sales:", maximum)
```

Question 6: You are analyzing product ratings:
ratings =

[[5, 4, 3],
[4, 4, 5],
[3, 5, 4]]

1. Compute average rating.
2. Print all ratings greater than 4.
3. Explain how nested loops are used in matrix-based ML computations.


```python
ratings =[[5, 4, 3], [4, 4, 5], [3, 5, 4]]
print(ratings)
```

```python
total = 0
count = 0

for row in ratings:
    for rating in row:
        total += rating
        count += 1

average = total / count

print("Average Rating:", average)
```

```python
print("Ratings greater than 4:")

for row in ratings:
    for rating in row:
        if rating > 4:
            print(rating)
```

A matrix is a table of rows and columns, similar to the ratings list above. Nested loops allow us to process every element in the matrix.


Applications in Machine Learning
1.Matrix Traversal
Access every feature value in a dataset.
Compute statistics such as sums, averages, and variances.
2.Matrix Multiplication
Neural networks perform matrix multiplications between inputs and weights.
Nested loops can be used to multiply rows and columns.
3.Distance Calculations
Algorithms like K-Nearest Neighbors compare data points feature by feature using loops.
4.Image Processing
Images are represented as matrices of pixel values.
Nested loops scan through rows and columns of pixels.

Question 8: Given
prices = [500, 1500, 2500, 800, 1200]
1. Use map() + lambda to add 18% GST to all prices.
2. Use filter() to select prices above 1000.
3. Explain where such transformations are used in real-world pipelines.

```python
prices = [500, 1500, 2500, 800, 1200]

gst_prices = list(map(lambda x: x * 1.18, prices))

print(gst_prices)
```

```python
prices = [500, 1500, 2500, 800, 1200]

filtered_prices = list(filter(lambda x: x > 1000, prices))

print(filtered_prices)
```

Real-World Use in Data Science Pipelines
1. Data Transformation (map)

map() is used to transform data before analysis or model training.

Examples:

Adding GST/tax to prices
Converting temperatures (°C → °F)
Scaling numerical features
Converting currencies

2. Data Filtering (filter)

filter() is used to remove unwanted records.

Examples:

Keeping customers with purchases above a threshold
Removing invalid data
Selecting high-value transactions
Filtering active users

3. Machine Learning Preprocessing

Before training a model:

Transform features using map()
Remove noisy or irrelevant records using filter()
Prepare clean, structured data for algorithms


Question 9:
Write a generator that yields squares of numbers from 1 to 1,000,000.
Explain difference between:
● List comprehension
● Generator expression
Why are generators preferred in big data scenarios?

```python
def square_generator():
    for i in range(1, 1000001):
        yield i ** 2
```

```python
squares = square_generator()

for _ in range(5):
    print(next(squares))
```
